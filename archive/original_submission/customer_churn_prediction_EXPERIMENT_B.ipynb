```python
def predict_customer_churn(customer_data):

    # Convert input into DataFrame
    customer_df = pd.DataFrame([customer_data])

    # Encode categorical features
    categorical_encoded = encoder.transform(
        customer_df[categorical_cols]
    )

    # Get numerical features
    numerical_values = customer_df[numerical_cols].values

    # Combine numerical + encoded categorical features
    customer_final = np.hstack([
        numerical_values,
        categorical_encoded
    ])

    # Make prediction
    prediction = xgb_model.predict(customer_final)[0]

    # Get probability
    probability = xgb_model.predict_proba(customer_final)[0][1]

    if prediction == 1:
        result = "Churn"
    else:
        result = "No Churn"

    return result, probability

print("Prediction function created successfully!")
```

```python
sample_customer = {
    "Gender": "Female",
    "Senior Citizen": "Yes",
    "Partner": "No",
    "Dependents": "No",
    "Tenure Months": 5,
    "Phone Service": "Yes",
    "Multiple Lines": "No",
    "Internet Service": "Fiber optic",
    "Online Security": "No",
    "Online Backup": "No",
    "Device Protection": "No",
    "Tech Support": "No",
    "Streaming TV": "Yes",
    "Streaming Movies": "Yes",
    "Contract": "Month-to-month",
    "Paperless Billing": "Yes",
    "Payment Method": "Electronic check",
    "Monthly Charges": 85.50,
    "Total Charges": 427.50
}

result, probability = predict_customer_churn(sample_customer)

print("Prediction:", result)
print("Churn Probability:", round(probability * 100, 2), "%")
```

```python
def predict_final_customer(customer_data):

    customer_df = pd.DataFrame([customer_data])

    categorical_encoded = encoder_final.transform(
        customer_df[categorical_cols]
    )

    numerical_values = customer_df[numerical_cols].values

    customer_final = np.hstack([
        numerical_values,
        categorical_encoded
    ])

    prediction = xgb_final.predict(
        customer_final
    )[0]

    probability = xgb_final.predict_proba(
        customer_final
    )[0][1]

    result = "Churn" if prediction == 1 else "No Churn"

    if probability >= 0.70:
        risk = "High Risk"
    elif probability >= 0.40:
        risk = "Medium Risk"
    else:
        risk = "Low Risk"

    return result, probability, risk
```
